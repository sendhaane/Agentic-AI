```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
```

GOOGLE API

```python


from langchain.chat_models import init_chat_model

model = init_chat_model("google_genai:gemini-3.5-flash-lite")

response = model.invoke ("what is your work")

print(response.content)
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

model = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")

response = model.invoke ("why do polar bear sleep a lot")

print(response.content)
```

GROQ API

```python
from langchain.chat_models import init_chat_model

model = init_chat_model("groq:openai/gpt-oss-120b")

response = model.invoke("what can you do")

print(response.content)
```

```python
from langchain_groq import ChatGroq

model = ChatGroq(model ="openai/gpt-oss-120b")

response = model.invoke("how did you get trained")

print(response.content)
```

```python
model.stream("how did you get trained")
```

```python
for chunk in model.stream("how did you get train"):
    print(chunk.text ,end="|")
```

```python
responces = model.batch(["why sky is blue",
                         "why water is blue",
                         "why leaves are in green color"],
                         config={
                             'max_concurrency':5,
                         }
                         )

for responce in responces:
    print(responce)
```
