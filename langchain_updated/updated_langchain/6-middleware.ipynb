Based on message lines

```python


model = init_chat_model(
    "groq:openai/gpt-oss-120b",
    temperature=0
)
agent = create_agent(
    model = model,
    checkpointer=InMemorySaver(),
    middleware=[
        SummarizationMiddleware(
            model="groq:openai/gpt-oss-20b",
            trigger=("messages",10),
            keep=("messages",4)
        )
    ]
    

)


config = {"configurable":{"thread_id":"test-1"}}

questions = [
    "what is 1",
    "what is 2",
    "what is 3",
    "what is 4",
    "what is 5",
    "what is 6",
    "what is 7",
    "what is 8",
    "what is 9",
    "what is 10",  
    "what is 11"     
]

for q in questions:
    responce = agent.invoke({"messages":[HumanMessage(content=q)]},
                            config
                            )
    print(responce)
    print(f"messages: {len(responce['messages'])}")
```

Humman in loop middleware


```python


from langchain_groq import ChatGroq
from langchain.tools import tool
from langchain.agents import create_agent
from langchain.agents.middleware import HumanInTheLoopMiddleware

from langgraph.checkpoint.memory import InMemorySaver
from langgraph.types import Command




model = ChatGroq(
    model="openai/gpt-oss-120b",
    temperature=0
)


@tool
def book_room(guest_name: str, date: str) -> str:
    """Book a hotel room for a guest on the requested date."""
    return f"Room successfully booked for {guest_name} on {date}."


memory = InMemorySaver()


agent = create_agent(
    model=model,
    tools=[book_room],
    system_prompt=(
        "You are a professional hotel booking assistant. "
        "Use the book_room tool when the guest requests a booking. "
        "Ask for missing information."
    ),
    middleware=[
        HumanInTheLoopMiddleware(
            interrupt_on={
                "book_room": {
                    "allowed_decisions": ["approve", "edit", "reject"]
                }
            }
        )
    ],
    checkpointer=memory
)


config = {
    "configurable": {
        "thread_id": "hotel_user_1"
    }
}


response = agent.invoke(
    {
        "messages": [
            {
                "role": "user",
                "content": "Book a room for Sendhaane on 2026-11-05."
            }
        ]
    },
    config=config
)


if response.get("__interrupt__"):

    print("\nHuman approval required!")

    for interrupt in response["__interrupt__"]:
        print(interrupt.value)

else:
    print(response["messages"][-1].content)

# 9. Resume after human approval
decision = input(
    "\nApprove booking? (yes/no): "
).strip().lower()

if decision == "yes":

    final_response = agent.invoke(
        Command(
            resume={
                "decisions": [
                    {"type": "approve"}
                ]
            }
        ),
        config=config
    )

    print("\nFinal response:")
    print(final_response["messages"][-1].content)

else:

    final_response = agent.invoke(
        Command(
            resume={
                "decisions": [
                    {
                        "type": "reject",
                        "message": "Booking rejected by the hotel manager."
                    }
                ]
            }
        ),
        config=config
    )

    print("\nFinal response:")
    print(final_response["messages"][-1].content)
```
