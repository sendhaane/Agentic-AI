```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
```

```python
from langchain_groq import ChatGroq

model = ChatGroq(model ="openai/gpt-oss-120b")

response = model.invoke("how many colors are seen in rainbow")

print(response.content)
```

```python
from langchain.tools import tool

@tool
def get_weather(location:str)->str:
    """Get the weather at loction"""
    return f"its rainy in {location}"

model_with_tool = model.bind_tools([get_weather])

```

```python
responce = model_with_tool.invoke("what is the weather in pondy")

print(responce)
```

```python
for tool_call in responce.tool_calls:
    print(f"Tool: {tool_call['name']}")
    print(f"Args: {tool_call['args']}")
```

```python
messages = [{"role":"user","content":"what is the weather in pondy"}]
ai_msg = model_with_tool.invoke(messages)
messages.append(ai_msg)
```

```python
for tool_call in responce.tool_calls:
    tool_result = get_weather.invoke(tool_call)
    print(tool_result)
    messages.append(tool_result)
```

```python
print(messages)
```

```python
final_response = model_with_tool.invoke(messages)
print(final_response)
```

```python
print(final_response.content)
```
